Write a Python function that calculates the mean of a matrix either by row or by column, based on a given mode. The function should take a matrix (list of lists) and a mode ('row' or 'column') as input and return a list of means according to the specified mode.

```python
#%pip install numpy 
import numpy as np
print(np.__version__)


def calculate_matrix_mean(matrix: list[list[float]], mode: str) -> list[float]:
    # Basic validations
    if not isinstance(matrix, list):
        raise TypeError("matrix must be a list of lists")
    if len(matrix) == 0:
        return []
    if any(not isinstance(row, list) for row in matrix):
        raise TypeError("matrix must be a list of lists")
    num_cols = len(matrix[0])
    if num_cols == 0:
        return []
    if any(len(row) != num_cols for row in matrix):
        raise ValueError("all rows must have the same length")

    means: list[float] = []

    if mode == "row":
        # Mean for each row
        for row in matrix:
            means.append(sum(row) / num_cols)

    if mode == "column":
        # Mean for each column
        num_rows = len(matrix)
        for j in range(num_cols):
            col_sum = 0.0
            for i in range(num_rows):
                col_sum += matrix[i][j]
            means.append(col_sum / num_rows)

    return means
```

```python
#Testing
m = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

print(calculate_matrix_mean(m, "row"))     # expected [2.0, 5.0, 8.0]
print(calculate_matrix_mean(m, "column"))  # expected [4.0, 5.0, 6.0]
```
